## **Complex Numbers**

### **What are complex numbers? How are they defined?**


Below are two graphs. The first graph shows the axis on which all real numbers lie. However, this does not mean that there are no numbers in the space above and below this axis - there are complex numbers. This is shown in the second graph below.

![https://flylib.com/books/2/729/1/html/2/images/0131089897/graphics/08fig01.gif](https://flylib.com/books/2/729/1/html/2/images/0131089897/graphics/08fig01.gif)

> Source: https://www.oreilly.com/library/view/understanding-digital-signal/0131089897/

Complex numbers are an extension of the real number system that makes it easy to model 2D space. Locating a point in 2D space using complex numbers is in many cases much more convenient than using the Cartesian coordinate system. Describing the mathematical and physical world accurately with complex numbers makes it much easier to perform certain calculations, since a number of mathematical operations can be performed on complex numbers.

To define a complex number, we need to determine its real and imaginary parts.

Let us define an example of a complex number:

* The real part tells us the horizontal position of the number. Let the real part of our number be 3.
* The imaginary part tells us whether our number is above or below the horizontal axis. Let the imaginary part of our number be 2.

In Python, we will define such a number as follows:

```python
a = 3 + 2j
```

![https://wumbo.net/concepts/complex-number-system/figures/complex-plane-650-400.light.svg](https://wumbo.net/concepts/complex-number-system/figures/complex-plane-650-400.light.svg)

> Source: https://wumbo.net/concepts/complex-numbers/

Our complex number $a$ is above the number 3 on the horizontal axis (we got this by declaring the real part of our number to be 3) and two units above the horizontal axis (we got this by declaring the imaginary part of our number to be 2). The visualization of complex numbers on the plane can take the form of an arrow (vector). It represents the direction and return of the shift of the zero number in such a way as to obtain our target number. That is, in the above case, to get the number $3 + 2i$, the number $0 + 0i$ must be shifted 3 units to the right and 2 units up - this rearranges the above vector.

The imaginary unit $i$ (coded as $j$ in Python) has its own mathematical definition:

$i^{2} = -1$

### **Trigonometric form of a complex number**

In some operations, it will be easier to describe a complex number by using the length of the aforementioned arrow (vector) $r$ and the angle between this arrow (vector) and the horizontal axis $\varphi$:

$z = x + yi = r(\cos\varphi + i\cdot \sin\varphi)$

![https://upload.wikimedia.org/wikipedia/commons/thumb/7/7a/Complex_number_illustration_modarg.svg/240px-Complex_number_illustration_modarg.svg.png](https://upload.wikimedia.org/wikipedia/commons/thumb/7/7a/Complex_number_illustration_modarg.svg/240px-Complex_number_illustration_modarg.svg.png)

> Source: https://en.wikipedia.org/wiki/Complex_number

The value of the parameter $r$ can be counted using the Pythagorean theorem:

$r = |z| = \sqrt{x^{2} + y^{2}}$

The angle between the arrow (vector) and the horizontal axis can be determined using trigonometric functions:

$ \begin{cases}
\cos\varphi = \frac{x}{r} \\
\sin\varphi = \frac{y}{r}
\end{cases} $
$\longrightarrow$
$\varphi (x, y) = $
$ \begin{cases}
\begin{matrix}
\arctan \big| \frac{y}{x} \big| & \text{x > 0 and y > 0} \\
\frac{1}{2}\pi & \text{x = 0 and y > 0} \\
\pi - \arctan \big| \frac{y}{x} \big| & \text{x < 0 and y > 0} \\
\pi & \text{x < 0 and y = 0} \\
\pi + \arctan \big| \frac{y}{x} \big| & \text{x < 0 and y < 0} \\
\frac{3}{2}\pi & \text{x = 0 and y < 0} \\
2\pi - \arctan \big| \frac{y}{x} \big| & \text{x > 0 and y < 0} \\
0 & \text{x = 0 and y = 0} \\
\end{matrix}
\end{cases} $

The following are defined functions that can be used to convert complex numbers from general form to trigonometric form (and vice versa):

```python
# Useful modules 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()
```

```python
def general2trigonometric(complex_number, deg_result = True, round_up = 3):
    
    x = np.real(complex_number)
    y = np.imag(complex_number)
    r = (x**2 + y**2)**(0.5)
    
    if x > 0 and y > 0:
        phi = np.arctan(np.abs(y/x))
    elif x == 0 and y > 0:
        phi = 0.5 * np.pi
    elif x < 0 and y > 0:
        phi = np.pi - np.arctan(np.abs(y/x))
    elif x < 0 and y == 0:
        phi = 1.0 * np.pi
    elif x < 0 and y < 0:
        phi = np.pi + np.arctan(np.abs(y/x))
    elif x == 0 and y < 0:
        phi = 1.5 * np.pi
    elif x > 0 and y < 0:
        phi = 2.0 * np.pi - np.arctan(np.abs(y/x))
    else:
        phi = 0
    
    if deg_result == True:
        return (np.round(r, round_up), np.round(np.rad2deg(phi), round_up))
    else:
        return (np.round(r, round_up), np.round(phi, round_up))
```

```python
def trigonometric2general(r, phi, deg_result = True, round_up = 2):
    
    if deg_result == True: 
        z = r * (np.cos(np.deg2rad(phi)) + 0+1j * np.sin(np.deg2rad(phi)))
    else:
        z = r * (np.cos(phi) + 1j * np.sin(phi))

    return (np.round(z.real, round_up) + 1j * np.round(z.imag, round_up))
```

Let's analyze the current problem with a concrete example. In the following, the angle of inclination of a vector and its length are calculated for an example complex number $2 + 1i$.

$z = 2 + 1i \\
\text{Re}(z) = x = 2 \\
\text{Im}(z) = y = 1$

![https://wumbo.net/concepts/complex-number-system/figures/complex-plane-1-2i-650-400.light.svg](https://wumbo.net/concepts/complex-number-system/figures/complex-plane-1-2i-650-400.light.svg)


$r = \sqrt{2^{2} + 1^{2}} \\
r = \sqrt{5} = 2.236$


$ \begin{cases}
\cos\varphi = \frac{2}{\sqrt{5}} \\
\sin\varphi = \frac{1}{\sqrt{5}} 
\end{cases}$
$\longrightarrow$
$\varphi = \arctan \big| \frac{1}{2} \big| = 26.565^{\text{o}}$

$z = 2.236\cdot(\cos(26.565^{\text{o}}) + i\cdot \sin(26.565^{\text{o}}))$

An analogous value is obtained using one of the functions defined above - *general2trigonometric*. Returning to the form containing the real and imaginary parts is easy, since it only requires calculating the individual values of the trigonometric functions and simplifying the notation. For verification, such a calculation was performed using the *trigonometric2general* function.

```python
general2trigonometric(2 + 1j)
```

```python
trigonometric2general(2.236, 26.565)
```

### **Addition of complex numbers**

In the following two complex numbers are declared. The result of the sum of these two complex numbers is stored in the variable 'c'.

```python
a = 2 + 3j
b = 4 - 1j

c = a + b
```

```python
print("Complex number 'a':", a)
print("Complex number 'b':", b)
print("__________________________")
print("Complex number 'c':", c)
```

$\text{Re}(a + b) = \text{Re}(a) + \text{Re}(b) = 2 + 4 = 6 \\
\text{Im}(a + b) = \text{Im}(a) + \text{Im}(b) = 3 + (-1) = 2$

### **Multiplication of complex numbers**

Two new complex numbers have been declared for further analysis. Python allows you to multiply complex numbers using the asterisk character - *.

```python
a = 4 + 3j
b = 2 + 1j

c = a * b
```

```python
print("Complex number 'a':", a)
print("Complex number 'b':", b)
print("__________________________")
print("Complex number 'c':", c)
```

In this case, it is not so easy to see how to perform multiplication of complex numbers. It is not as intuitive as in the case of addition, where it was enough to add the real and imaginary parts separately. The matter becomes clear when we switch to the trigonometric form. For this, the previously defined function *general2trigonometric* was used:

```python
a = general2trigonometric(a)
b = general2trigonometric(b)

c = general2trigonometric(c)
```

```python
print("Complex number 'a' - trig. form:", a)
print("Complex number 'b' - trig. form:", b)
print("_________________________________________________")
print("Complex number 'c' - trig. form:", c)
```

Now it is much easier to see the connections:

$r_{a\cdot b} = r_{a}\cdot r_{b} = 5 \cdot 2.236 = 11.18 \\
\varphi_{a\cdot b} = \varphi_{a} + \varphi_{b} = 36.87^{\text{o}} + 26.565^{\text{o}} = 63.435^{\text{o}}$

In this way, we also obtained the geometric interpretation of the multiplication of two complex numbers. It consists in scaling (multiplying) the value of the length of the vector of the first complex number by the value of the length of the second vector. The resulting vector will indicate a point on the complex plane, i.e. the value of the multiplication of these numbers, if this rescaled vector is further subjected to rotation by an angle between the horizontal axis and the vector of the second complex number. This is well illustrated in the following figure.

![https://wumbo.net/concepts/complex-number-system/figures/complex-multiplication-angles-and-radii-540-420.light.svg](https://wumbo.net/concepts/complex-number-system/figures/complex-multiplication-angles-and-radii-540-420.light.svg)

Of course, if all you care about is the result, the multiplication of the complex numbers shown above can be done analytically:

$(4 + 3i)(2+1i) = 8 + 4i + 6i + 3i^{2} = 8 + 4i + 6i + 3 \cdot (-1) = 8 + 4i + 6i - 3 = 5 + 10i$

### **Exponentiation of complex numbers**

Exponentiation is the multiplication of the same value many times. In this case, we will analyze a case very similar to the one above.

```python
a = 1 + 1j
c = a * a * a
```

```python
print("Complex number 'a':", a)
print("Complex number 'a':", a)
print("Complex number 'a':", a)
print("__________________________")
print("Complex number 'c':", c)
```

By analogy with multiplication, we will switch to trigonometric form:

```python
a = general2trigonometric(a)
c = general2trigonometric(c)
```

```python
print("Complex number 'a' - trig. form:", a)
print("Complex number 'a' - trig. form:", a)
print("Complex number 'a' - trig. form:", a)
print("_________________________________________________")
print("Complex number 'c' - trig. form:", c)
```

$r_{a^{3}} = (r_{a})^{3} = 1.414^{3} = 2.828 \\
\varphi_{a^{3}} = 3\cdot \varphi_{a} = 3\cdot 45^{\text{o}} = 135^{\text{o}}$

$r_{a^{n}} = (r_{a})^{n} \\
\varphi_{a^{n}} = n\cdot \varphi_{a}$

The geometric interpretation of the multiplication of complex numbers is no different from that of the multiplication of two numbers. The only difference is that a vector is scaled and rotated several times. It is worth mentioning a formula that uses the above relationship and allows you to quickly calculate the result of the triplication of complex numbers in trigonometric form. We are talking about De Moivre's formula:

$z^{n} = r^{n}(\cos(n \cdot \varphi) + i\cdot \sin(n \cdot \varphi))$

### **Roots of complex numbers**

The situation is even worse for roots of complex numbers. A considerable number of publicly available functions do not handle the calculation of roots of complex numbers.

```python
z = 2 + 3j
# s = z^(1/2)    # Error!
# s = np.sqrt(z) # Incomplete answer
```

However, knowing that the root can be written as a power with an exponent that is a rational number, nothing prevents you from using the de Moivre formula shown above to create your own function. Let's remember the periodicity of trigonometric functions:

$\sqrt[n]{z} = z^{\frac{1}{n}} = r^{\frac{1}{n}}\left(\cos\left(\frac{\varphi+2k\pi}{n}\right)+i\sin\left(\frac{\varphi+2k\pi}{n}\right)\right),\quad k\in\{0,\dots, n-1\}$

```python
def sqrt_complex(z, n = 2, round_up = 5):
    
    r = general2trigonometric(z, deg_result = 0, round_up = 5)[0]
    phi = general2trigonometric(z, deg_result = 0, round_up = 5)[1]
    
    results = []
    for k in range(0, n, 1):
        period = 2 * k * np.pi
        sqrt_z = r**(1/n) * ((np.cos((phi + period)/n)) + 1j*(np.sin((phi + period)/n)))      
        results = np.append(results, (np.round(sqrt_z.real, round_up) + 1j*(np.round(sqrt_z.imag, round_up))))

    return results
```

Listed below are the results of calculations of various roots of complex numbers. A short comment has been added to each of them. Examples are given from the simplest to the most difficult calculations.

* $\sqrt{4} = s $

```python
s = sqrt_complex(4 + 0j)
print(s)
```

In the first example we see the results of finding the value of $\sqrt{4}$. Every natural number is also a complex number, so the *sqrt_complex* function will return the correct values of the roots of this number, i.e. $s\in\{-2, 2\}$. Check and verify the correctness of the results for this example below:

$(-2) \cdot (-2) = 4 \\
2 \cdot 2 = 4$

* $\sqrt{-4} = s$

```python
s = sqrt_complex(-4 + 0j)
print(s)
```

The next example is not so obvious. The $\sqrt{-4}$ had no solutions in the set of real numbers. No number multiplied by itself could return a negative value. However, it turns out that this square root has solutions! The *sqrt_complex* function returned two results: $s\in\{-2i, 2i\}$. Both results are numbers from the set of complex numbers. Check and verify the correctness of the results for the second example below:

$2i \cdot 2i = 4 \cdot i^{2} = 4 \cdot (-1) = -4 \\
(-2i) \cdot (-2i) = 4 \cdot i^{2} = 4 \cdot (-1) = -4$

* $\sqrt[4]{2 + 2i} = s$

```python
z = 2 + 2j
n = 4

s = sqrt_complex(z, n)
print(s)
```

When determining the square root of $n^{th}$, we should get $n$ solutions. This is also the case when determining the value of $\sqrt[4]{2 + 2i}$. The *sqrt_complex* function has returned four solutions. To get a better idea of the relationship between them, the graph below shows the results of this square root on the complex plane.

```python
plt.figure(figsize = (8, 8))
plt.scatter(s.real, s.imag, label = f'Results of rooting the {n}-th degree of the number {z}', color = '#c32148')
plt.title('Root results of a complex number represented in the complex plane')
plt.xlabel('Re()')
plt.ylabel('Im()')
plt.xlim([-2, 2])
plt.ylim([-2, 2])
plt.legend()
plt.show()
```

The results of rooting complex numbers are plotted on a circle. The radius of this circle is always $\sqrt[n]{r}$. Below is a function that instantly visualizes the results of rooting any complex number you specify.

```python
def sqrt_complex_plot(z, n = 2):
    
    s = sqrt_complex(z, n)
    angle = np.linspace(0, 2 * np.pi, 150) 
    r = ((z.real**2 + z.imag**2)**(0.5))**(1/len(s))
 
    plt.figure(figsize = (8, 8))
    plt.scatter(z.real, z.imag, label = f'Complex number {z}', color = '#073763')
    plt.quiver([0], [0], [z.real], [z.imag], units = 'xy', scale = 1, color = '#839bb1')
    plt.scatter(s.real, s.imag, label = f'Results of rooting the {n}-th degree of the number {z}', color = '#c32148')
    plt.plot(r * np.cos(angle), r * np.sin(angle), color = '#bcbcbc')

    if z.real >= z.imag:
        plt.xlim([-1.2*z.real, 1.2*z.real])
        plt.ylim([-1.2*z.real, 1.2*z.real])
    else:
        plt.xlim([-1.2*z.imag, 1.2*z.imag])
        plt.ylim([-1.2*z.imag, 1.2*z.imag])
    plt.title('Root results of a complex number represented in the complex plane')
    plt.xlabel('Re()')
    plt.ylabel('Im()')
    plt.legend()

    return plt.show()
```

```python
sqrt_complex_plot(z, n)
```

On the complex plane, all roots are equally spaced on the circle. In addition, there will always be a root result on the complex plane that has an argument equal to $\varphi_{s_{1}} = \frac{1}{n}\varphi_{z}$. This will be clearly visible when the results are transformed into trigonometric form:

```python
# Argument of the first root result (in degrees)
general2trigonometric(s[0])[1]
```

```python
# 1/n of the argument of the number from which the root was counted (in degrees)
(1/n) * general2trigonometric(z)[1]
```
